```python
pipeline_table=f"{catalog}.{pipeline_schema}.pipeline_control"
```

```python
dbutils.widgets.text("batch_id", "") 
batch_id = dbutils.widgets.get("batch_id")
```

Mark the current batch as "in_progress" by appending a new status row to pipeline_control

```python
from pyspark.sql import Row
from pyspark.sql import functions as F

if batch_id:
    in_progress_df = (
        spark.createDataFrame(
            [Row(batch_id=batch_id, status="in_progress")]
        )
        .withColumn("created_timestamp", F.current_timestamp())
        .withColumn("updated_timestamp", F.current_timestamp())
    )

    (
        in_progress_df.write
            .format("delta")
            .mode("append")
            .saveAsTable(pipeline_table)
    )

    print(f"Batch # {batch_id} is in_progress")
else:
    raise Exception("batch_id is missing")  
```
